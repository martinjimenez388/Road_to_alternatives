Now see how many NaN elements are for every column in the dataset. The output is a Series with the column names as indices and the number of NaN elements as values. datfile.isna() by itself only returns a DataFrame of the same shape as datfile, but with True or False values depending on whether the element is NaN or not. The sum() method then counts the number of True values for each column.

```python
nans = datfile.isna().sum()
nans[nans > 0]
#nans[nans > 0].size
```

Extract the rows corresponding to cars coming from Asia. From it extract the feature "fuel_efficiency_mpg", and then find its maximum.

```python
np.max(datfile[datfile["origin"]=="Asia"].fuel_efficiency_mpg)
```

Drop the NaN entries of a column and compute the average.

```python
filtered_HP = datfile["horsepower"].dropna()
np.average(filtered_HP)
```

Extract the most frequent value and how many times it appears.

```python
mstfreq = filtered_HP.value_counts().index[0]
filtered_HP.value_counts().head(n=1)
```

Fill the missing values in the "horsepower" column with the most frequent value from the previous step.

```python
datfile["horsepower"] = datfile["horsepower"].fillna(mstfreq)
```

```python
datfile["horsepower"].value_counts().head(n=1)
```

Compute again the average.

```python
np.average(datfile["horsepower"].values)
```

From the rows corresponding to Asian cars, keep the columns "vehicle_weight" and "model_year".

```python
asiancars = datfile[datfile["origin"]=="Asia"]
X = asiancars[["vehicle_weight", "model_year"]].head(n=7).values
```

```python
Xm1 = np.linalg.inv(X.T @ X)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = Xm1 @ X.T @ y
np.sum(w)
```
