Drop the NaN entries of a column and compute the average.

```python
filtered_HP = datfile["horsepower"].dropna()
np.average(filtered_HP)
```

Extract the most frequent value and how many times it appears.

```python
mstfreq = filtered_HP.value_counts().index[0]
filtered_HP.value_counts().head(n=1)
```

Fill the missing values in the "horsepower" column with the most frequent value from the previous step.

```python
datfile["horsepower"] = datfile["horsepower"].fillna(mstfreq)
```

```python
datfile["horsepower"].value_counts().head(n=1)
```

Compute again the average.

```python
np.average(datfile["horsepower"].values)
```

<span style="color: orange">Q7. Sum of weights.</span>
<span style="color: lightgreen">Ans.: 0.369</span>

Now, from the rows corresponding to Asian cars, keep the columns "vehicle_weight" and "model_year".

```python
asiancars = datfile[datfile["origin"]=="Asia"]
X = asiancars[["vehicle_weight", "model_year"]].head(n=7).values
```

```python
Xm1 = np.linalg.inv(X.T @ X)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = Xm1 @ X.T @ y
np.sum(w)
```
